# Estudio reproducible de siete defectos

UCI Steel Plates Faults, DOI 10.24432/C5J88N, CC BY 4.0. No hay clase sana. Separación estratificada 60/20/20; selección solo en validación. Ejecutar desde la raíz mediante scripts/check_notebook.py.

```python
from pathlib import Path
import pandas as pd
from steel_faults.training import train
from steel_faults.inference import Predictor
report = train(Path("data/raw/Faults.NNA"), Path("var/notebook-model"))
pd.DataFrame({name: {"macro_f1": r["macro_f1"], "accuracy": r["accuracy"]} for name, r in report["test"].items()}).T
```

```python
pd.DataFrame(report["test"][report["selected"]]["per_class"]).T.loc[report["classes"]]
```

```python
sample = pd.DataFrame([report["examples"][0]["features"]])
predictor = Predictor("var/notebook-model")
result = predictor.predict(sample)[0]
assert result["score"] == max(predictor.pipeline.predict_proba(sample)[0])
result
```

La equivalencia usa exactamente la misma entrada física. Las puntuaciones no están calibradas. La partición aleatoria no acredita generalización por lote, planta o tiempo; se requieren datos externos para decisiones operacionales. Los costos de aceptar piezas defectuosas no se pueden medir sin una clase sana y costos del proceso.
